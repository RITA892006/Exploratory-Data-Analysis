```python
import numpy as np
```

```python
np.random.seed(1)
arr=np.random.randint(10,size=(2,5))
print(arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
print(arr.itemsize)
print(arr.nbytes)
```

```python
print(arr[1,2])
```

```python
np.random.seed(1)
arr=np.random.randint(10,size=5)
print(arr)
print(arr[-2])
print(arr[2])
print(arr[3])
print(arr[2:4])
print(arr[1:3])
print(arr[1:4:2])
print(arr[::-1])
```
