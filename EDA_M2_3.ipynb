```python
#50
import pandas as pd

data = {
    'department': ['CSE', 'AIML', 'ECE', 'CSE', 'AIML', 'ECE',
                   'CSE', 'AIML', 'ECE', 'CSE', 'AIML', 'ECE'],
    'gender': ['M', 'F', 'M', 'F', 'M', 'F',
               'M', 'F', 'M', 'F', 'M', 'F'],
    'salary': [60000, 65000, 55000, 62000, 70000, 58000,
               68000, 72000, 61000, 64000, 75000, 60000]
}

df = pd.DataFrame(data)

print("Employee DataFrame:")
print(df)

pivot_table = pd.pivot_table(
    df,
    values='salary',
    index='gender',
    columns='department',
    aggfunc='mean'
)

print("\nAverage Salary Pivot Table:")
print(pivot_table)
```

```python
#51
import pandas as pd
pivot_count = pd.pivot_table(
    df,
    values='salary',
    index='gender',
    columns='department',
    aggfunc='count'
)

print(pivot_count)
```

```python
#52
pivot_totals = pd.pivot_table(
    df,
    values='salary',
    index='gender',
    columns='department',
    aggfunc='mean',
    margins=True,
    margins_name='Total'
)

print(pivot_totals)
```

```python
#53
import pandas as pd

df['age'] = [22, 25, 28, 30, 35, 40,
             24, 27, 32, 38, 42, 45]

# Create age groups
df['age_group'] = pd.cut(
    df['age'],
    bins=[20, 29, 39, 50],
    labels=['Junior', 'Mid', 'Senior']
)

pivot_age = pd.pivot_table(
    df,
    values='salary',
    index='age_group',
    columns='department',
    aggfunc='mean'
)

print(pivot_age)
```

```python
#54
pivot_multi = pd.pivot_table(
    df,
    values=['salary', 'age'],
    index='department',
    aggfunc={
        'salary': 'sum',
        'age': 'mean'
    }
)

print(pivot_multi)
```
