```python
import numpy as np
arr = np.arange(1, 21)

print("Array:", arr)
print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Data type:", arr.dtype)
```

```python
np.random.seed(0)  
arr = np.random.randint(0, 51, (4, 5))

print("Array:\n", arr)
print("Item size:", arr.itemsize)
print("Total bytes:", arr.nbytes)
```

```python
arr = np.arange(1, 13).reshape(3, 4)
print("Array:\n", arr)

print("Row 2, Col 3:", arr[1, 2])

print("Last row:", arr[-1])

print("Last column:", arr[:, -1])
```

```python
print("First 2 rows & columns:\n", arr[:2, :2])
print("Every alternate column:\n", arr[:, ::2])
print("Columns reversed:\n", arr[:, ::-1])
```

```python
big = np.arange(20).reshape(5, 4)
sub = big[:3, :4] 
sub[0, 0] = 999
print("Original affected:\n", big)

copy_sub = big[:3, :4].copy()
copy_sub[0, 0] = 111
print("Original unchanged:\n", big)
```

```python
x = np.arange(12)
m1 = x.reshape(3, 4)
col1 = x.reshape(12, 1)
col2 = x[:, np.newaxis]

print("Are both column versions same?", np.array_equal(col1, col2))
```

```python
a = np.random.randint(0, 10, (2, 3))
b = np.random.randint(0, 10, (2, 3))

print("Vertical stack:\n", np.vstack((a, b)))
print("Horizontal stack:\n", np.hstack((a, b)))
```

```python
arr = np.arange(15)
parts = np.split(arr, 3)

for i, p in enumerate(parts):
    print("Part", i + 1, ":", p)
```

```python
arr = np.arange(24).reshape(6, 4)

print("Horizontal split (two 6×2 arrays):")
print(np.hsplit(arr, 2))

print("Vertical split (two 3×4 arrays):")
print(np.vsplit(arr, 2))
```

```python

x = np.random.randint(0, 100, 10)
print("Original array:", x)
sorted_x = np.sort(x)
print("Sorted (np.sort):", sorted_x)
print("Original unchanged:", x)
x.sort()
print("Sorted (x.sort):", x)
```

```python
x = np.random.randint(0, 100, 10)
print("Original array:", x)
i = np.argsort(x)
print("Indices:", i)

print("Sorted using indices:", x[i])
print("Same as np.sort(x):", np.sort(x))
```

```python
arr = np.random.randint(0, 100, (4, 6))
print("Original array:\n", arr)
row_sorted = np.sort(arr, axis=1)
print("Rows sorted:\n", row_sorted)

col_sorted = np.sort(arr, axis=0)
print("Columns sorted:\n", col_sorted)
```

```python
x = np.random.randint(0, 100, 15)
print("Original array:", x)
part = np.partition(x, 4)
print("Partitioned array:", part)
print("4 smallest values:", part[:4])
print("Sorted array:", np.sort(x))
```

```python
student_names = np.array(["Asha", "Ravi", "Meena", "Kiran", "Sita"])
scores = np.array([85, 72, 90, 60, 78])

# Sort by scores
order = np.argsort(scores)
print("Students ranked by score:")
for i in order:
    print(student_names[i], "-", scores[i])
```
