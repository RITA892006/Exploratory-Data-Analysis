```python
import pandas as pd
import numpy as np

data = pd.Series([1, np.nan, 'hello', None])
print("Series:\n", data)

print("\nIs Null:\n", data.isnull())

# Filter non-null values
print("\nNon-null values:\n", data[~data.isnull()])
```

```python
df = pd.DataFrame(np.arange(20).reshape(5,4))
df.iloc[0,1] = np.nan
df.iloc[2,3] = np.nan
df.iloc[4,0] = np.nan
df.iloc[3,2] = np.nan

print("DataFrame:\n", df)
print("\nMissing values per column:\n", df.isnull().sum())
```

```python
print("Drop rows with NaN:\n", df.dropna())
print("\nDrop columns with NaN:\n", df.dropna(axis='columns'))
```

```python
df2 = pd.DataFrame({
    'A':[1, np.nan, 3, None],
    'B':[np.nan, np.nan, 7, None],
    'C':[np.nan, np.nan, 9, None]
})
df2.loc[1] = [np.nan, np.nan, np.nan]  # fully empty row

print("Original:\n", df2)
print("\nDrop only fully empty rows:\n", df2.dropna(how='all'))
```

```python
print("Keep rows with at least 3 non-NaN:\n", df2.dropna(thresh=3))
```

```python
import pandas as pd
import numpy as np

s = pd.Series([1, np.nan, 2, None, 3, np.nan])

print("Fill with 0:\n", s.fillna(0))
print("\nForward fill:\n", s.ffill())
print("\nBackward fill:\n", s.bfill())

```

```python
temps = pd.DataFrame({
    'Day': range(1,8),
    'Temp': [30, np.nan, 32, 31, np.nan, 33, 34]
})

print("Original:\n", temps)

# Fill with mean
temps_mean = temps.copy()
temps_mean['Temp'] = temps_mean['Temp'].fillna(temps_mean['Temp'].mean())
print("\nFill with mean:\n", temps_mean)

# Linear interpolation
temps_interp = temps.copy()
temps_interp['Temp'] = temps_interp['Temp'].interpolate()
print("\nLinear interpolation:\n", temps_interp)
```

```python
tuples = [
    ('Karnataka', 2010), ('Karnataka', 2020),
    ('Maharashtra', 2010), ('Maharashtra', 2020),
    ('Tamil Nadu', 2010), ('Tamil Nadu', 2020)
]
index = pd.MultiIndex.from_tuples(tuples)
pop = pd.Series([61095297, 67562686, 112374333, 123144223, 72147030, 75695000], index=index)
print(pop)
```

```python
print("\nAll values for Karnataka:\n", pop['Karnataka'])
print("\nAll values for year 2020:\n", pop.loc[(slice(None), 2020)])
print("\nSingle value (Maharashtra, 2010):\n", pop[('Maharashtra', 2010)])
```

```python
df_pop = pop.unstack()
print("\nDataFrame:\n", df_pop)

print("\nBack to Series:\n", df_pop.stack())
```

```python
index = pd.MultiIndex.from_product([['Math','Science'], ['Alice','Bob']], names=['Subject','Student'])
columns = pd.MultiIndex.from_product([['Term1','Term2'], ['Mid','Final']], names=['Term','Exam'])
marks = pd.DataFrame(np.random.randint(50,100,(4,4)), index=index, columns=columns)
print(marks)
```

```python

print(marks.loc['Math'])
print(marks.T.groupby(level='Term').mean().T)


```
